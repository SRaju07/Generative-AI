# 5. ReAct Prompting - (Reasoning + Acting) AI reasons through a problem, takes actions using tools, observes the results, and generates an answer:

```python
import os
import json
from groq import Groq

# Add your NEW Groq API key here
GROQ_API_KEY = "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"

# Initialize Groq client:
client=Groq(api_key=GROQ_API_KEY)

# Select model:
MODEL="openai/gpt-oss-120b"

# Get User question:
question=input("Enter your query : ")

# Create prompt:
prompt = f"""
    <role>
        You are an expert Technical Assistant.
    </role>

    <task>
        Solve the user's problem accurately and efficiently.
    </task>

    <process>
        1. Understand the user's problem.
        2. Identify the important information.
        3. Determine the best solution.
        4. Use available tools when necessary.
        5. Observe and evaluate the tool results.
        6. Verify the solution.
        7. Provide the final answer.
    </process>

    <guidelines>
        - Use simple and clear English.
        - Explain technical concepts step by step.
        - Provide examples or code when relevant.
        - Do not assume missing information.
        - Do not reveal hidden chain-of-thought.
        - Provide only a brief reasoning summary.
    </guidelines>

    <user_question>
        {question}
    </user_question>

    <output_format>
        Reasoning Summary:
        Provide a short explanation of the approach.

        Action:
        Mention any tools used, if applicable.
        
        Observation:
        Summarize the relevant results, if applicable.
        
        Final Answer:
        Provide the clear and accurate solution.
        </output_format>
"""



# Generate AI response:
response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "user", "content": prompt}
    ]
)

# Display response:
print("\n========== AI RESPONSE ==========\n")
print(response.choices[0].message.content)
```

# 8. Prompt Chaining - Prompt Chaining connects multiple prompts so the output of one step becomes the input for the next step:

```python
from groq import Groq

# STEP 1: Add your NEW Groq API key
GROQ_API_KEY = "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"

# STEP 2: Initialize Groq client
client = Groq(api_key=GROQ_API_KEY)

# STEP 3: Select model
MODEL = "openai/gpt-oss-120b"

# STEP 4: Get user question
question = input("Enter your query: ")

# =============================
# CHAIN 1: Generate Explanation
# =============================

prompt_1 = f"""
Explain the following user question.

Provide:
- Key concepts
- Important points
- Relevant examples

Do not write the final polished answer yet.

User Question:
{question}
"""

response_1 = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "user", "content": prompt_1}
    ]
)

intermediate_result = response_1.choices[0].message.content

print("\n========== CHAIN 1 OUTPUT ==========\n")
print(intermediate_result)

# ============================
# CHAIN 2: Create Final Answer
# ============================

prompt_2 = f"""
Create a clear and simple final answer using the
information generated by the first prompt.

Original User Question:
{question}

Information from Chain 1:
{intermediate_result}

Instructions:
- Give only the final answer.
- Keep it short and simple.
- Maximum 5 sentences.
"""

response_2 = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "user", "content": prompt_2}
    ]
)

# ============================
# DISPLAY FINAL RESPONSE
# ============================

print("\n========== FINAL AI RESPONSE ==========\n")
print(response_2.choices[0].message.content)
```

# 9. Dynamic Prompt Templates - you can write a prompt once and use it many times by changing only the input value :

```python

from groq import Groq

# STEP 1: Add your NEW Groq API key
GROQ_API_KEY = "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"

# STEP 2: Initialize Groq client
client = Groq(api_key=GROQ_API_KEY)

# STEP 3: Select model
MODEL = "openai/gpt-oss-120b"

# STEP 4: Get user question
question = input("Enter your question: ")

# STEP 5: Select response style
style = input(
    "Enter Response Style "
    "(summarize / translate / explain / code): "
).strip().lower()

# STEP 6: Define dynamic prompt templates
templates = {
    "summarize": "Summarize the following text in 3 concise bullet points.",

    "translate": "Translate the following text into French.",

    "explain": "Explain the following concept clearly in simple English with examples.",

    "code": "Write only a short and working Python code for the following task. Do not explain the code."
}

# STEP 7: Select the appropriate template
selected_template = templates.get(
    style,
    templates["explain"]
)

# STEP 8: Create dynamic prompt
prompt = f"""
You are an expert AI assistant.

Task:
{selected_template}

User Question:
{question}

Follow the selected response style carefully.
"""

# STEP 9: Generate AI response
response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "user", "content": prompt}
    ]
)

# STEP 10: Display response
print("\n========== AI RESPONSE ==========\n")
print(response.choices[0].message.content)
```
